```python
class LinkedList(object):
  def __init__(self):
    self.head = None
    self.tail = None

  def print(self):
    currNode = self.head
    while(currNode.next):
      print(currNode.value, end=" -> ")
      currNode = currNode.next
    print(currNode.value)
    
  def append(self, value):
    new_node = Node(value)
    if self.head == None:
      self.head = new_node
      self.tail = new_node
    else:
      self.tail.next = new_node
      self.tail = self.tail.next

  def get(self, idx):
    current = self.head
    for _ in range(idx):
      current = current.next
    return current.value
  
  def insert(self, idx, value):
    current = self.head
    next = self.head.next

    for _ in range(idx):
      current = current.next
      next = current.next

    current.next = Node(value)
    current.next.next = next


  def delete(self, idx):
    current = self.head

    for _ in range(idx - 1):
      current = current.next

    current.next = current.next.next
    pass
```
