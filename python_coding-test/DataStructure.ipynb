## 메모리?

그럼 우리의 데이터를 정하는 공간은 어디일까? `메모리`이다.

메모리는 HDD(하드 디스크), RAM(램 메모리) 두 가지가 있다.

코드는 HDD에, 데이터는 RAM 메모리에 올라가게 된다.

비효율적인 자료구조를 사용하면 RAM 메모리 낭비를 초래하게 되고, 실행 시간이 더 느려질 수 있다. 

RAM은 전기 신호를 저장할 수 있는 작은 반도체 소자인 트랜지스터로 구성되어 있다.

이진수, 즉 0과 1을 표현할 수 있고 이걸 표현하는 칸을 1 bit(binary Digit)라고 한다.

1 bit - 2가지, 2 bit - 4가지의 수를 표현할 수 있다.

\+ 8bit = 1byte이다.

더 큰 수를 표현하려면 16진수로 표현하는 방법도 있다.

만약 217을 2진수, 16진수로 변환한다고 생각하면..

2진수 : 11011001

16진수 : D9

16진수 앞에는 0x를 붙여서 16진수임을 표현한다.

만약 1MB 램을 사용한다고 치면

1Byte x 1024 = 1KB,
1KB x 1024 = 1MB

하나 하나의 바이트에 주소를 단다고 생각했을 때
총 2의 20승의 주소를 저장할 수 있다.

## 메모리 할당

### int

4 byte를 차지함

### Character

컴퓨너틑 기본적으로 문자열을 메모리에 저장할 수 없음.

그래서 문자를 숫자로 저장하기로 표현한 것이 아스키 코드이다.

1 byte를 차지함

### List

```python
int array[4] = {1, 2, 3, 4}
```

Array의 경우 저장된 데이터 값이 차지하는 byte를 따른다.

위에 예시의 경우 4 byte x 4 = 16 byte를 차지하는 것이다.

### LinkedList

LinkedList는 Array와는 다르게 다음 값을 가르키는 주소도 같이 저장하기 때문에 value와 address가 합쳐진 하나의 Node로 표현한다.

int 형으로 저장하기 때문에 4byte(value) + 4byte(address) = 8byte로 표현한다.
