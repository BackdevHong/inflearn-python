```python
# 접근 방법을 직관적으로 생각하기
```

```python
# nums = [4, 1, 9, 7, 5, 3, 16]
# target = 14

nums = [2, 1, 5, 7]
target = 4

def TwoSum(nums, target):
  n = len(nums)
  for i in range(n):
    for j in range(i + 1, n):
      if nums[i] + nums[j] == target:
        return True
  return False

TwoSum(nums, target)
```

```python
# 접근 방식을 자료 구조와 알고리즘을 활용하는 방향으로
# Two Pointer , Sort
```

```python
def TwoSum(nums, target):
  nums.sort()
  n = len(nums)
  l, r = 0, n - 1

  while l < r:
    if nums[l] + nums[r] > target:
      r -= 1
    elif nums[l] + nums[r] < target:
      l += 1
    elif nums[l] + nums[r] == target:
      return True
  return False

nums = [2, 1, 5, 7]
target = 4

# nums = [4, 1, 9, 7, 5, 3, 16]
# target = 14

TwoSum(nums, target)
```
