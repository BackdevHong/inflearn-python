# 알고리즘
- 문제를 해결하는 일련의 `절차`나 `방법`
- 자주 쓰이는 문제 해결 방법은 패턴화
  - ex) BFS, DFS, Binary Search, Dijkstra 등
- 하나의 문제를 해결할 수 있는 알고리즘은 다양하다.
  - 각 문제에 적합한 알고리즘을 선택할 수 있어야 한다.
  - 알고리즘을 평가할 수 있어야 한다.

## 평가 기준
- 시간 복잡도(Time Complexity)
- 공간 복잡도(Space Complexity)
- 구현 복잡도
- `우리나라같은 경우 공간 복잡도는 거의 보지 않는다.`

시간 복잡도와 공간 복잡도는 trade-off 상태
- 실행 시간을 줄이면, 메모리를 더 사용해야 하고
- 메모리를 줄이면, 실행 시간이 늘어나게 된다.

우리는 공간 복잡도는 제쳐두고, 시간 복잡도를 우선으로 생각한다.

## Running Time

만약 공이 5개가 있다고 치고, 해당 공을 작은 숫자부터 큰 숫자로 정렬을 하려고 한다.

( 5, 3, 2, 1, 4 )

우리가 정렬을 하는 방식으로는

5개의 공 중 가장 작은 것을 맨 앞에 두고, 다시 나머지 4개의 공 중 가장 작은 것을 맨 앞에 주는 방식으로 진행한다.

이것을 코드로 만들면 다음과 같다.

```python
def selectionSort(arr, length):
  n = length

  for i in range(n - 1):
    min_idx = i
    for j in range(i + 1, n):
      if (arr[j] < arr[min_idx]):
        min_idx = j

    arr[i], arr[min_idx] = arr[min_idx], arr[i]
    print(arr)
  return arr

selectionSort([1, 5, 2, 4, 1], 5)
```

그런데, 이 연산을 하는데 실행 시간은 얼마나 걸릴까?

프로그램을 실행하면 CPU가 코드를 한줄, 한줄 처리하게 되는데, 시작하는 시점부터 모든 줄을 처리하는데 걸리는 시간을 Running Time 이라고 한다.

반복문이 있는 코드도 보면 반복 횟수를 곱해줘야 하기 때문에 반복이 10배가 늘어다면 실행 횟수도 10배가 늘어나게 된다.

이와 같이 시간과 입력값 n의 관계를 시간 복잡도라고 한다.

## BIG - O

시간 복잡도를 간단히 표기하기 위해 시간 복잡도의 최고차수 n만을 사용해 표현하는 점근 표기법이다.

### O(1)

T(n) = 530, T(n) = 260, T(n) = 1

위와 같은 복잡도가 있다고 쳤을 때 시간복잡도가 n이라고 쳤을 때 n이 커지면 커질 수록 차이가 미미해진다.

따라서 상수 시간 복잡도는 O(1)으로 점근표기한다.

### O(N)

T(n) = 5n + 30
T(n) = 5n
T(n) = n

위와 같은 복잡도가 있다고 쳤다고 쳤을 때 n2(제곱)과 n의 차이가 심하게 커지기 때문에 위와 같은 시간 복잡도는 O(n) 으로 표기할 수 있다.

## 시간 복잡도 순서

O(n!) > O(2n) > O(n2) > O(n log n) > O(n) > O(log n) > O(1)

## Case

Case는 3가지가 존재한다.
- Base Case
- Average Case
- Worst Case

우리가 아까 만들었던 삽입 정렬의 시간 복잡도를 구해보자.

```python
def selectionSort(arr, length):
  n = length

  for i in range(n - 1):
    min_idx = i
    for j in range(i + 1, n):
      if (arr[j] < arr[min_idx]):
        min_idx = j

    arr[i], arr[min_idx] = arr[min_idx], arr[i]
    print(arr)
  return arr

selectionSort([1, 5, 2, 4, 1], 5)
```

n번 반복하는 반복문이 두개가 중첩되어 있기에 O(n2)이지만 들어오는 데이터에 따라 시간 복잡도가 달라진다.

이미 정렬되어 있는 형태로 들어온다면 반복문이 가장 적게 들어와서 시간 복잡도가 적게 돌아가는 한편,
역순으로 정렬되어 있는 형태라면 반복문이 가장 많이 돌아가 시간 복잡도가 많이 늘어난다.


이렇게 입력 값의 크기가 동일하다 하더라도, 형태에 따라서 시간 복잡도가 달라진다.

이렇게 나온 시간 복잡도에서...

가장 높은 시간 복잡도 : Worst Case

가장 낮은 시간 복잡도 : Best Case

평균 시간 복잡도 : Average Case
