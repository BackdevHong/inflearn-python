```python
# Design Browser History
# 인터넷 브라우저에서 방문 기록을 구현하라!
```

```python
# 제약조건
# 1 <= homepage.length <= 20
# 1 <= url.length <= 20
# 1 <= step <= 100
# homepage와 url은 '.'을 포함한 lower case 영어 문자로 구성되어 있다.
# visit, back 그리고 forward는 최대 5000번의 호출이 있을 수 있다.
```

```python
# 내가 구현한 버전
```

```python
class Node(object):
  def __init__(self, value=0, next=None, prev=None):
    self.value = value
    self.next = next
    self.prev = prev  
    pass

class BrowserHistory(object):
  def __init__(self, homepage):
    new_node = Node(homepage)
    
    self.head = new_node
    self.tail = new_node
    self.now = new_node

  def visit(self, url : str) -> None:
    nextNode = Node(url)

    if (self.tail == self.head):
      nextNode.prev = self.head
      self.head.next = nextNode

      self.tail = nextNode
      self.now = nextNode
    else:
      self.now.next = nextNode
      nextNode.prev = self.now
      self.now = self.now.next

      self.now = self.now

  def back(self, steps: int) -> str:
    current = self.now
    for _ in range(steps):
      if (current.prev == None):
        break
      else:
        current = current.prev
    self.now = current
    return current.value

  def forward(self, steps: int) -> str:
    current = self.now
    for _ in range(steps):
      if (current.next == None):
        break
      else:
        current = current.next
    self.now = current
    return current.value
```

```python
class ListNode(object):
  def __init__(self, val = 0, next=None, prev=None):
    self.val = val
    self.next = next
    self.prev = prev

class BrowserHistory(object):
  def __init__(self, homepage):
    self.head = self.current = ListNode(val=homepage)

  def visit(self, url):
    self.current.next = ListNode(val=url, prev=self.current)
    self.current = self.current.next
    return None

  def back(self, steps):
    while steps > 0 and self.current.prev != None:
      steps -= 1
      self.current = self.current.prev
    return self.current.val

  def forward(self, steps):
    while steps > 0 and self.current.next != None:
      steps -= 1
      self.current = self.current.next
    return self.current.val
```
