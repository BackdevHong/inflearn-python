```python
animals = []
```

```python
animals.append('코알라')
animals.append('하이에나')
animals.append('바다소')
animals.append('땅다람쥐')
animals.append('스컹크')
animals.append('바다코끼리')
animals.append('아나콘다')
```

```python
for animal in animals:
  print(animal)
```

```python
for num in [1, 2, 3]:
  print(num)
```

```python
for my_str in "김왼손의 왼손코딩":
  print(my_str)
```
